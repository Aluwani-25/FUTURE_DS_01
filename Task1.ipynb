```python
import pandas as pd
import numpy as np

df= pd.read_csv("online_retail.csv")
df
```

```python
df.head()
```

```python
df= df.drop_duplicates()
```

```python
df = df.dropna(subset=['CustomerID','Description'])
```

```python
df
```

```python
df['InvoiceDate']= pd.to_datetime(df['InvoiceDate'])
df['CustomerID']= df['CustomerID'].astype('Int64')
```

```python
returns= df['InvoiceNo'].str.startswith('C',na=False)
df_returns=df[returns]
df= df[~returns]
```

```python
df = df[(df['UnitPrice']>0) &(df['Quantity']>0)]
```

```python
df['Description'] = df['Description'].str.strip().str.upper()
```

```python
df['Revenue']=df['UnitPrice']*df['Quantity']
```

```python
df['Year']= df['InvoiceDate'].dt.year
```

```python
df
```

```python
df['UnitPrice'] = (
    df['UnitPrice']
    .astype(str)
    .str.strip()
    .str.replace(',', '.', regex=False)   # fix decimal commas
    .str.replace(r'[^0-9\.]', '', regex=True)  # remove currency symbols
)
```

```python
df['UnitPrice'] = pd.to_numeric(df['UnitPrice'], errors='coerce')
```

```python
print(df['UnitPrice'].dtype)
```

```python
df['Revenue'] = (
    df['Revenue']
    .astype(str)
    .str.strip()
    .str.replace(',', '.', regex=False)   # fix decimal commas
    .str.replace(r'[^0-9\.]', '', regex=True)  # remove currency symbols
)
```

```python
df['Revenue'] = pd.to_numeric(df['Revenue'], errors='coerce')
```

```python
df['Month']= df['InvoiceDate'].dt.month
```

```python
df['Day']= df['InvoiceDate'].dt.day
```

```python
df['Revenue'] = df['Revenue'].round(2)
```

```python
# Make sure all types are correct
df['UnitPrice'] = pd.to_numeric(df['UnitPrice'], errors='coerce')
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
df['Revenue'] = pd.to_numeric(df['Revenue'], errors='coerce')
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'], errors='coerce')

# Optional: drop bad rows
df = df.dropna()

```

```python
df.to_json("online_retail_cleaned.json", orient='records', lines=True)
```

```python
df.to_csv("online_retail_cleaned.csv", index=False, encoding='utf-8')
```
